```python
## load the pickle file

import pickle
import tensorflow as tf
from tensorflow.keras.models import load_model
import numpy as np
import pandas as pd

```

```python
## load the trained model, scale and onehot encoder objects file


model  = load_model('model2.keras')


## load the encoder and scaler objects

with open('onehot_encoder_geo.pkl', 'rb') as file:
    label_encoder_geo = pickle.load(file)
                                
with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
                       
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender = pickle.load(file)  
```

```python
## example input data for prediction
input_data = {
  "CreditScore": 650,
  "Geography": "France",
  "Gender": "Male",
  "Age": 38,
  "Tenure": 5,
  "Balance": 75000.50,
  "NumOfProducts": 2,
  "HasCrCard": 1,
  "IsActiveMember": 1,
  "EstimatedSalary": 92000.00
}
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
## one hot encode Geography

geo_encoded = label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded, columns=label_encoder_geo.get_feature_names_out(['Geography']))  
geo_encoded_df
```

```python
input_data = pd.concat([input_df.reset_index(drop=True), geo_encoded_df.reset_index(drop=True)], axis=1)
input_data 
```

```python
## encode gender variable

input_df["Gender"] =label_encoder_gender.transform(input_df["Gender"])
input_df
```

```python
## concatenate encoded geography columns with input data

input_df = pd.concat([input_df.drop("Geography", axis=1), geo_encoded_df], axis=1)
input_df
```

```python
## scaling the input data

input_Scaled = scaler.transform(input_df)
```

```python
## predict churn

prediction = model.predict(input_Scaled)
print(prediction)
```
